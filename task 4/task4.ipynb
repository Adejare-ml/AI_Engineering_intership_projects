```python
import pandas as  pd 
import numpy  as np
%matplotlib inline
import matplotlib.pyplot as  plt 


df  =pd.read_csv(r"C:\Users\adelu\OneDrive\Documents\AI_Engineering_intership_projects\task 4\car-sales-extended-missing-data.csv" , encoding ="latin-1")
df
```

```python
# Show original columns
print("Original columns:", df.columns.tolist())

# Normalize columns: lower case, strip whitespace, replace spaces with underscores, and remove parentheses
df.columns = (
    df.columns.str.strip()
    .str.lower()
    .str.replace(' ', '_', regex=False)
    .str.replace('(', '', regex=False)
    .str.replace(')', '', regex=False)
)
print("Normalized columns:", df.columns.tolist())
```

```python
# 1. Summary statistics for numerical and categorical columns
display(df.describe(include='all'))

# 2. Count of missing values per column
print("\nMissing values per column:\n", df.isnull().sum())

# 3. Value counts for categorical columns (including NaNs)
for col in df.select_dtypes(include=['object']).columns:
    print(f"\n--- Value counts for {col} ---")
    print(df[col].value_counts(dropna=False))
```

```python
# Handle numerical missing values (Impute with Median)
for col in df.select_dtypes(include=['float64', 'int64']).columns:
    df[col] = df[col].fillna(df[col].median())

# Handle categorical missing values (Impute with 'missing' string)
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].fillna('missing')

# Verify no missing values remain
print("Remaining missing values:\n", df.isnull().sum())
```
