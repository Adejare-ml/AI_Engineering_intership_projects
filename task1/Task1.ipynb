```python
# imports
import pandas as pd
import  requests
```

```python
# fetching  the  article  for use
url = "http://mlg.ucd.ie/files/datasets/bbc-fulltext.zip"

response = requests.get(url)
# downloading  the  BBC News articles zip file
with open("bbc.zip", "wb") as f:
    f.write(response.content)

print("download ready")
```

```python
#  the  file  is  zipped  so we  need  unzip it  using  this  library
import zipfile

with zipfile.ZipFile("bbc.zip", 'r') as zip_ref:
    zip_ref.extractall(r"C:\Users\adelu\OneDrive\Documents\AI_engineering_intership")
```

```python
# Now  we call  all  the files  using  (pathlib)
from pathlib import Path

# lets  look at the sub folders 
p = Path(r'C:\Users\adelu\OneDrive\Documents\AI_engineering_intership\bbc')
[x for x in p.iterdir() if x.is_dir()]

# listing all the  individual text  files  in the  folder
list(p.glob('**/*.txt'))
```

```python
# adding all the text  files  to  a  list 
articles = []

for filepath in p.glob('**/*.txt'):
    if filepath.parent != p:  # update to remove  the readme file from  being  added
        with open(filepath, "r", encoding="utf-8", errors="ignore") as f:
            text = f.read()
            articles.append(text)
```

```python
# checking  total length for the  list 
print(len(articles))
print(articles[0][:200])
```

```python
# converting the  list  into  a Dataframe
df = pd.DataFrame(articles,columns=["Original_text"])
df.head()
```

```python
# make the all  the  text in the  'Original_text' lowercase for  easy  opeation 
df["lowercase_Original_text"] = df["Original_text"].str.lower()
df
```

```python
# removing punctuation from  lowercase_orginal_text
df["No_punctuation"]  =  df["lowercase_Original_text"].str.replace("[^a-zA-Z0-9\s]" ,"" ,regex=True)
df
```

```python
# Checks 
df[["Original_text", "No_punctuation"]].head(2) 
```

```python
# Tokenization 
df ["tokens"] = df["No_punctuation"].str.split()
df.head(3)
```

```python
# import  nltk 
import nltk
nltk.download("stopwords")
from nltk.corpus import stopwords
stop_words = set(stopwords.words("english"))
```

```python
# removing  stopwords using  (nltk) natural language   toolkit 
df["tokens_clean"] = df["tokens"].apply(lambda token_list: [word for word in token_list if word not in stop_words])
df.head(2)
```

```python
# Save  to CSv  we  convert  to  from token list  back  to  a  string 
df["cleaned_df"] = df["tokens_clean"].apply(lambda tokens_list: " ".join(tokens_list))
df
```

```python
# save to csv
df[["Original_text", "cleaned_df"]].to_csv("bbc_cleaned.csv", index=False)
```
